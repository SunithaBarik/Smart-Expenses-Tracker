```python
# 💰 Smart Expense Tracker - By Sunitha Barik
import pandas as pd
import matplotlib.pyplot as plt

expenses = []

print("Namaste Sunitha! Expense Tracker Ready hai! 💰")
```

```python
# Step 2 - Kharcha add karne wala function
def add_expense(item, amount, category):
    expenses.append({
        "Item": item,
        "Amount": amount,
        "Category": category
    })
    print(f"✅ {item} - Rs.{amount} ({category}) add ho gaya!")

# Test karte hai - 3 kharche
add_expense("Chai", 20, "Khana")
add_expense("Shopping", 500, "Shopping")
add_expense("Metro", 100, "Travel")

print("\nTotal kharche:", len(expenses))
```

```python
# Step 3 - Table banao aur Total nikalo
df = pd.DataFrame(expenses)

print("--- Mera Kharcha Table ---")
print(df)

total = df["Amount"].sum()
print(f"\n💸 Total Kharcha: Rs.{total}")

print("\n--- Category Wise ---")
print(df.groupby("Category")["Amount"].sum())
```

```python
# Step 4 - Chart banao
category_total = df.groupby("Category")["Amount"].sum()

plt.figure(figsize=(8, 4))

# Pie Chart
plt.subplot(1, 2, 1)
plt.pie(category_total, labels=category_total.index, autopct='%1.1f%%')
plt.title("Category Wise Kharcha")

# Bar Chart
plt.subplot(1, 2, 2)
plt.bar(category_total.index, category_total.values, color=['orange','pink','skyblue'])
plt.title("Kharcha Comparison")
plt.ylabel("Rupees")

plt.tight_layout()
plt.show()

print("🎉 Project Complete Sunitha! Tumne Expense Tracker bana liya!")
```
